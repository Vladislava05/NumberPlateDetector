```python
from ultralytics import YOLO
```

```python
model = YOLO('yolov8n.pt')
results = model.train(
    data='/content/Gujarat_Vehicle_Dataset/data.yaml',
    epochs=10,
    imgsz=640,
    batch=16)
```

```python
import cv2
import os
from google.colab.patches import cv2_imshow
image_path = '/content/Gujarat_Vehicle_Dataset/images/val/IMG_20260226_123415.jpg'
image = cv2.imread(image_path)
results = model(image)
annotated_frame = results[0].plot()
output_path = 'detected_image.jpg'
cv2.imwrite(output_path, annotated_frame)
cv2_imshow(annotated_frame)
for result in results:
    for box in result.boxes:
        print(f"Обнаружен объект: {result.names[box.cls[0].item()]} с уверенностью {box.conf[0].item():.2f}")
```
